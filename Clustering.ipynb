# Clustering

# 1. What is unsupervised learning in Machine Learning?

Unsupervised learning finds patterns or groups in data without labeled output.

Examples:
- K-Means
- Hierarchical Clustering
- DBSCAN


# 2. How does K-Means clustering work?

Steps:
1. Choose K (number of clusters).
2. Initialize K centroids.
3. Assign each data point to the nearest centroid.
4. Recalculate centroids.
5. Repeat until clusters become stable.


# 3. What is a dendrogram in hierarchical clustering?

A dendrogram is a tree-like diagram that shows how data points or clusters
are merged or separated during hierarchical clustering.

It helps decide the number of clusters.


# 4. K-Means vs Hierarchical Clustering

K-Means:
- Requires K in advance.
- Faster for large datasets.
- Produces fixed K clusters.

Hierarchical:
- Does not require K initially.
- Produces a dendrogram.
- Usually slower for large datasets.


# 5. Advantages of DBSCAN over K-Means

- Does not require number of clusters in advance.
- Can find irregular-shaped clusters.
- Detects noise/outliers.
- Works well with clusters of different shapes.


# 6. When do you use Silhouette Score?

Silhouette Score is used to evaluate the quality of clustering.

It measures:
- How similar a point is to its own cluster.
- How far it is from other clusters.

Higher score = better-defined clusters.


# 7. Limitations of Hierarchical Clustering

- Computationally expensive for large datasets.
- Sensitive to noise and outliers.
- Once clusters are merged/split, the decision cannot easily be reversed.
- Results depend on the linkage method.


# 8. Why is feature scaling important in K-Means?

K-Means uses distance calculations.

If features have different scales, a large-scale feature can dominate the
distance calculation.

Scaling gives features a comparable influence.


# 9. How does DBSCAN identify noise points?

DBSCAN uses two parameters:

- eps: Neighborhood radius.
- min_samples: Minimum points required in the neighborhood.

A point that does not have enough neighboring points and is not reachable
from a core point is considered noise.


# 10. What is inertia in K-Means?

Inertia is the sum of squared distances between each data point and its
cluster centroid.

Lower inertia generally means tighter clusters.


# 11. What is the Elbow Method?

The Elbow Method helps choose the optimal K in K-Means.

Steps:
1. Run K-Means with different K values.
2. Calculate inertia for each K.
3. Plot K vs inertia.
4. Choose the point where the decrease starts slowing significantly.

This point is called the "elbow".


# 12. What is "density" in DBSCAN?

Density means the number of data points present within a specified
neighborhood.

In DBSCAN, a region is dense when enough points are found within the
eps radius.


# 13. Can hierarchical clustering be used on categorical data?

Yes, but standard Euclidean distance is not suitable for categorical data.

Use appropriate distance measures such as:
- Hamming distance
- Jaccard distance

Categorical data may also need suitable encoding depending on the method.


# 14. What does a negative Silhouette Score indicate?

A negative Silhouette Score indicates that a data point may be assigned to
the wrong cluster.

It is closer to another cluster than its assigned cluster.


# 15. What is linkage criteria in hierarchical clustering?

Linkage criteria defines how the distance between two clusters is calculated.

Common types:
- Single linkage → minimum distance.
- Complete linkage → maximum distance.
- Average linkage → average distance.
- Ward linkage → minimizes within-cluster variance.


# 16. Why can K-Means perform poorly with varying cluster sizes or densities?

K-Means assumes roughly spherical and similarly sized clusters.

It can incorrectly assign points when:
- Cluster sizes are very different.
- Cluster densities are very different.
- Clusters have irregular shapes.


# 17. What are the core parameters of DBSCAN?

Two main parameters:

1. eps
   - Defines the neighborhood radius.

2. min_samples
   - Defines the minimum number of points needed to form a dense region.

Small eps/min_samples → more points may become noise.

Large eps/min_samples → larger or fewer clusters may be formed.


# 18. How does K-Means++ improve standard K-Means initialization?

K-Means++ chooses initial centroids more intelligently.

It selects centroids that are spread apart instead of choosing them randomly.

Benefits:
- Better initial clusters.
- Faster convergence.
- More stable results.
- Often better final clustering.


# 19. What is Agglomerative Clustering?

Agglomerative clustering is a bottom-up hierarchical clustering method.

Steps:
1. Start with each data point as its own cluster.
2. Merge the closest clusters.
3. Continue merging.
4. Stop when the desired number of clusters is reached.


# 20. Why is Silhouette Score better than only inertia?

Inertia only measures how compact the clusters are.

Silhouette Score considers both:
- Cluster cohesion → points should be close to their own cluster.
- Cluster separation → points should be far from other clusters.

Therefore, Silhouette Score gives a better indication of whether clusters are
both compact and well-separated.

# 11. Overlapping Blobs + K-Means + Decision Boundaries
# ============================================================

X, y = make_blobs(n_samples=500, centers=3,
                  cluster_std=2.5, random_state=42)

model = KMeans(n_clusters=3, random_state=42, n_init=10)
labels = model.fit_predict(X)

x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1

xx, yy = np.meshgrid(
    np.linspace(x_min, x_max, 300),
    np.linspace(y_min, y_max, 300)
)

grid = np.c_[xx.ravel(), yy.ravel()]
Z = model.predict(grid).reshape(xx.shape)

plt.contourf(xx, yy, Z, alpha=0.2)
plt.scatter(X[:, 0], X[:, 1], c=labels)
plt.scatter(model.cluster_centers_[:, 0],
            model.cluster_centers_[:, 1],
            marker='X', s=200)

plt.title("K-Means with Decision Boundaries")
plt.show()

# 30. QUICK SUMMARY OF IMPORTANT PARAMETERS
# ============================================================

# K-Means:
# n_clusters → Number of clusters
# n_init → Number of centroid initializations

# DBSCAN:
# eps → Neighborhood radius
# min_samples → Minimum points required for a dense region

# Agglomerative Clustering:
# n_clusters → Number of final clusters
# linkage → single / complete / average / ward

# PCA:
# n_components → Number of dimensions after reduction

# t-SNE:
# n_components=2 → Reduce to 2D for visualization
# perplexity → Controls local/global neighborhood balance

# Silhouette Score:
# Range = -1 to 1
# Higher score → Better separated and compact clusters

# K-Means Inertia:
# Lower inertia → More compact clusters
