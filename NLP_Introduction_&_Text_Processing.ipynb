# ============================================================
# Q1. What is Computational Linguistics and how does it relate
#     to NLP?
# ============================================================

# Computational Linguistics (CL) is the study of how computers
# can process, understand, and generate human language.
#
# NLP (Natural Language Processing) applies computational
# linguistics, machine learning, and deep learning techniques
# to real-world language problems.
#
# In simple terms:
# Computational Linguistics -> linguistic theories + computation
# NLP -> practical applications of language processing
#
# Examples: speech recognition, translation, chatbots,
# sentiment analysis, text summarization.


# ============================================================
# Q2. Historical evolution of Natural Language Processing
# ============================================================

# 1. 1950s-1960s:
#    Rule-based systems and linguistic rules were used.
#    Example: machine translation systems.
#
# 2. 1970s-1980s:
#    Knowledge-based and symbolic NLP systems became popular.
#    Systems used grammars, dictionaries, and manually created rules.
#
# 3. 1990s:
#    Statistical NLP became popular.
#    Machine learning models learned patterns from large text datasets.
#    Examples: Hidden Markov Models and statistical language models.
#
# 4. 2000s:
#    More data and better ML algorithms improved NLP.
#    Methods such as TF-IDF, Naive Bayes, and SVM were widely used.
#
# 5. 2010s:
#    Deep learning transformed NLP.
#    Word embeddings, RNNs, LSTMs, CNNs, and sequence-to-sequence
#    models became common.
#
# 6. 2018-Present:
#    Transformer architectures and large language models became dominant.
#    Examples: BERT, GPT, T5, and other foundation models.

# ============================================================
# Q4. What is text normalization and why is it essential?
# ============================================================

# Text normalization is the process of converting text into a
# consistent and standardized form before further processing.
#
# Common normalization steps:
# - Convert text to lowercase
# - Remove unnecessary punctuation
# - Remove extra spaces
# - Expand contractions
# - Handle special characters
# - Remove or replace URLs
# - Stemming or lemmatization
#
# Example:
# "HELLO!!!   Running Quickly."
#
# After normalization:
# "hello running quickly"
#
# Importance:
# - Reduces unnecessary variations in text
# - Improves model performance
# - Reduces vocabulary size
# - Makes tokenization and analysis easier


# ============================================================
# Q5. Stemming vs Lemmatization
# ============================================================

# Stemming:
# - Removes prefixes/suffixes using simple rules.
# - May produce words that are not actual dictionary words.
#
# Example:
# "playing" -> "play"
# "studies" -> "studi"
# "running" -> "run"
#
# Lemmatization:
# - Converts a word to its meaningful dictionary base form.
# - Uses vocabulary and linguistic information.
#
# Example:
# "playing" -> "play"
# "studies" -> "study"
# "better" -> "good" (depending on POS/context)
#
# Comparison:
#
# Stemming:
#   Faster
#   Simpler
#   Less accurate
#
# Lemmatization:
#   More accurate
#   More computationally expensive
#   Produces meaningful words
#
# Stemming is useful when speed is important.
# Lemmatization is preferred when linguistic correctness matters.

# Q6. Extract all email addresses using Regular Expressions
# ============================================================

import re

text = """
Hello team, please contact us at support@xyz.com for technical issues,
or reach out to our HR at hr@xyz.com. You can also connect with John
at john.doe@xyz.org and jenny via jenny_clarke126@mail.co.us.
For partnership inquiries, email partners@xyz.biz.
"""

pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b'

emails = re.findall(pattern, text)

print("Email addresses:")
for email in emails:
    print(email)

# Output:
# support@xyz.com
# hr@xyz.com
# john.doe@xyz.org
# jenny_clarke126@mail.co.us
# partners@xyz.biz

# Q7. String Tokenization and Frequency Distribution using NLTK
# ============================================================

import nltk
from nltk.tokenize import word_tokenize
from nltk.probability import FreqDist
import string

nltk.download('punkt')
nltk.download('punkt_tab')

paragraph = """
Natural Language Processing (NLP) is a fascinating field that combines
linguistics, computer science, and artificial intelligence. It enables
machines to understand, interpret, and generate human language.
Applications of NLP include chatbots, sentiment analysis, and machine
translation. As technology advances, the role of NLP in modern solutions
is becoming increasingly critical.
"""

# Tokenization
tokens = word_tokenize(paragraph)

print("All tokens:")
print(tokens)

# Convert to lowercase
tokens_lower = [token.lower() for token in tokens]

# Remove punctuation
words = [
    token for token in tokens_lower
    if token not in string.punctuation
]

# Frequency distribution
freq_dist = FreqDist(words)

print("\nFrequency Distribution:")
print(freq_dist)

# Most common words
print("\n10 Most Common Words:")
print(freq_dist.most_common(10))

# Q9. Word2Vec using Gensim
# ============================================================

# Install if required:
# !pip install gensim

from gensim.models import Word2Vec
from gensim.utils import simple_preprocess

dataset = [
    "Natural language processing enables computers to understand human language",
    "Word embeddings are a type of word representation that allows words with similar meaning to have similar representation",
    "Word2Vec is a popular word embedding technique used in many NLP applications",
    "Text preprocessing is a critical step before training word embeddings",
    "Tokenization and normalization help clean raw text for modeling"
]

# Tokenization and preprocessing
sentences = [
    simple_preprocess(sentence, deacc=True)
    for sentence in dataset
]

print("Preprocessed Dataset:")
for sentence in sentences:
    print(sentence)

# Train Word2Vec model
model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4,
    sg=1,          # Skip-gram
    epochs=100
)

# Get vector for a word
word = "language"

print("\nVector for 'language':")
print(model.wv[word])

# Find similar words
print("\nSimilar words to 'language':")
print(model.wv.most_similar("language", topn=5))

# Note:
# This is a very small dataset, so the learned embeddings are
# only for demonstration. Real Word2Vec models require much
# larger datasets to learn meaningful semantic relationships.
