# Regression

1. What is Simple Linear Regression?

Simple Linear Regression is a Machine Learning method used to predict a dependent variable (Y) using one independent variable (X).

Equation:
Y = mX + c


2. What are the key assumptions of Simple Linear Regression?

- Linear relationship between X and Y.
- Independent observations.
- Constant variance of errors (homoscedasticity).
- Errors are normally distributed.
- No significant outliers.


3. What does the coefficient m represent in Y = mX + c?

m is the slope/coefficient.

It represents how much Y changes when X increases by 1 unit.

Example:
m = 5 → Y increases by 5 when X increases by 1.


4. What does the intercept c represent?

c is the intercept.

It represents the predicted value of Y when X = 0.


5. How do we calculate the slope m?

Formula:

m = Σ[(X - X̄)(Y - Ȳ)] / Σ[(X - X̄)²]


6. What is the purpose of the least squares method?

It finds the best-fitting regression line by minimizing the sum of squared differences between actual and predicted values.

Error = Actual Y - Predicted Y


7. How is R² interpreted?

R² (R-squared) shows how much variation in the dependent variable is explained by the model.

Range: 0 to 1

Example:
R² = 0.80 → Model explains 80% of the variation in Y.


8. What is Multiple Linear Regression?

Multiple Linear Regression predicts a dependent variable using two or more independent variables.

Equation:

Y = b0 + b1X1 + b2X2 + ... + bnXn


9. Main difference between Simple and Multiple Linear Regression

Simple Linear Regression:
- One independent variable.

Multiple Linear Regression:
- Two or more independent variables.


10. What are the key assumptions of Multiple Linear Regression?

- Linear relationship.
- Independent observations.
- Homoscedasticity.
- Normally distributed residuals.
- No severe multicollinearity.
- No influential outliers.


11. What is heteroscedasticity?

Heteroscedasticity occurs when the variance of residuals is not constant.

Effect:
- Standard errors can become unreliable.
- Statistical tests and confidence intervals may be incorrect.


12. How can you improve a model with high multicollinearity?

- Remove highly correlated features.
- Combine related features.
- Use VIF to identify multicollinearity.
- Use Ridge or Lasso regression.
- Apply dimensionality reduction such as PCA.


13. Techniques for transforming categorical variables

- One-Hot Encoding.
- Label Encoding.
- Ordinal Encoding.
- Target Encoding.


14. What is the role of interaction terms?

Interaction terms allow the model to capture situations where the effect of one variable depends on another variable.

Example:

Y = b0 + b1X1 + b2X2 + b3(X1 × X2)


15. How can intercept interpretation differ between Simple and Multiple Regression?

Simple Regression:
- c is the predicted Y when the single X = 0.

Multiple Regression:
- c is the predicted Y when all independent variables = 0.


16. What is the significance of the slope?

The slope shows the direction and amount of change in Y for a one-unit change in X.

Positive slope → Y increases.
Negative slope → Y decreases.
Larger absolute slope → Greater change in Y.


17. How does the intercept provide context?

The intercept gives the expected value of Y when all predictor variables are zero.

It provides a baseline for interpreting the regression relationship.


18. Limitations of using R² alone

- High R² does not guarantee a good model.
- It does not prove causation.
- It can increase when unnecessary variables are added.
- It does not show prediction error directly.
- It does not detect all model problems.


19. What does a large standard error mean?

A large standard error means the estimated coefficient is less precise.

It indicates greater uncertainty about the coefficient value.


20. How can heteroscedasticity be identified in residual plots?

Plot residuals against predicted values.

Homoscedasticity:
- Random spread with roughly constant width.

Heteroscedasticity:
- Funnel or cone-shaped pattern.

It should be addressed because it can make standard errors and statistical inference unreliable.


21. High R² but low adjusted R² — what does it mean?

It usually means the model has unnecessary predictors.

R² increases when variables are added, while Adjusted R² penalizes unnecessary variables.


22. Why scale variables in Multiple Linear Regression?

Scaling puts variables on a similar scale.

Benefits:
- Makes coefficients easier to compare.
- Helps numerical optimization.
- Important when using regularization such as Ridge and Lasso.

Note:
Scaling is not strictly required for ordinary Linear Regression.


23. What is Polynomial Regression?

Polynomial Regression models a nonlinear relationship by adding powers of the independent variable.

Example:

Y = b0 + b1X + b2X²


24. How does Polynomial Regression differ from Linear Regression?

Linear Regression:
Y = b0 + b1X

Polynomial Regression:
Y = b0 + b1X + b2X² + b3X³

Polynomial Regression can model curved relationships.


25. When is Polynomial Regression used?

It is used when the relationship between variables is nonlinear but can be represented by a polynomial curve.

Example:
- Growth trends.
- Temperature vs energy consumption.
- Sales vs advertising.


26. General equation for Polynomial Regression

Y = b0 + b1X + b2X² + b3X³ + ... + bnXⁿ


27. Can Polynomial Regression be applied to multiple variables?

Yes.

Polynomial features can be created from multiple variables.

Example:

Y = b0 + b1X1 + b2X2 + b3X1² + b4X2² + b5X1X2


28. Limitations of Polynomial Regression

- High-degree polynomials can overfit.
- Can become complex.
- Sensitive to outliers.
- Higher degrees can cause unstable predictions.
- Interpretation becomes difficult.


29. Methods to evaluate model fit when selecting polynomial degree

- R².
- Adjusted R².
- MAE.
- MSE.
- RMSE.
- Cross-validation.
- Validation/test-set performance.

Cross-validation is generally preferred for selecting the degree.


30. Why is visualization important in Polynomial Regression?

Visualization helps us:

- See the relationship between X and Y.
- Check whether the curve fits the data.
- Detect underfitting.
- Detect overfitting.
- Identify outliers.
