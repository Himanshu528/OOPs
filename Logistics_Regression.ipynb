# Logistics Regression

1. What is Simple Linear Regression (SLR)? Explain its purpose.

Simple Linear Regression is a Machine Learning/statistical method used to predict a dependent variable (Y) using one independent variable (X).

Purpose:
- Find the relationship between X and Y.
- Predict Y based on X.


2. What are the key assumptions of Simple Linear Regression?

- Linear relationship between X and Y.
- Observations are independent.
- Residuals have constant variance (Homoscedasticity).
- Residuals are approximately normally distributed.
- No significant outliers.


3. Write the mathematical equation and explain each term.

Y = mX + c

Where:
Y = Predicted/Dependent variable
X = Independent variable
m = Slope/Coefficient
c = Intercept

Example:
Y = 2X + 5

Here:
m = 2
c = 5


4. Real-world example of Simple Linear Regression

Predict salary based on years of experience.

X = Years of experience
Y = Salary

Example:
More experience → Higher expected salary.


5. What is the method of least squares?

Least Squares is a method used to find the best-fitting regression line.

It minimizes the sum of squared differences between actual and predicted values.

Error = Actual value - Predicted value


6. What is Logistic Regression? How does it differ from Linear Regression?

Logistic Regression is used mainly for classification problems.

Example:
- Spam / Not Spam
- Pass / Fail
- Yes / No

Difference:

Linear Regression:
- Predicts continuous numerical values.
- Example: Salary, house price.

Logistic Regression:
- Predicts class/probability.
- Example: 0 or 1.


7. Three common evaluation metrics for regression

1. MAE (Mean Absolute Error)
- Average absolute difference between actual and predicted values.
- Lower MAE → Better model.

2. MSE (Mean Squared Error)
- Average squared difference between actual and predicted values.
- Lower MSE → Better model.

3. RMSE (Root Mean Squared Error)
- Square root of MSE.
- Lower RMSE → Better model.


8. What is the purpose of R-squared?

R² measures how much variation in the dependent variable is explained by the regression model.

Range:
0 to 1

Example:
R² = 0.80

→ The model explains approximately 80% of the variation in Y.


9. Python code to fit Simple Linear Regression

from sklearn.linear_model import LinearRegression

X = [[1], [2], [3], [4], [5]]
y = [2, 4, 6, 8, 10]

model = LinearRegression()

model.fit(X, y)

print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)


10. How do you interpret the coefficients?

Slope (m):
- Shows how much Y changes when X increases by 1 unit.
- Positive slope → Y increases.
- Negative slope → Y decreases.

Intercept (c):
- Represents the predicted value of Y when X = 0.

Example:

Y = 3X + 10

Slope = 3
→ Y increases by 3 for every 1-unit increase in X.

Intercept = 10
→ When X = 0, predicted Y is 10.
