# ============================================================
# QUESTION 2
# Perceptron: Architecture, Working and Limitations
# ============================================================

# Basic architecture:
#
# Inputs → Weighted Sum → Activation Function → Output
#
# x1 ──┐
# x2 ──┼──> Σ(w*x) + b ──> Activation ──> Output
# x3 ──┘

# Working:
# 1. Takes input features.
# 2. Multiplies each input by its weight.
# 3. Adds bias.
# 4. Applies an activation function.
# 5. Produces the output.

# Formula:
# output = activation(w1*x1 + w2*x2 + ... + b)

# Limitations:
# - Can solve only linearly separable problems.
# - Cannot solve XOR using a single perceptron.
# - Cannot learn complex nonlinear relationships.


# ============================================================
# QUESTION 3
# Activation Functions
# ============================================================

# Activation functions introduce non-linearity into neural networks.
# Without them, multiple layers would behave like a linear model.

# Sigmoid:
# Range: 0 to 1
# Commonly used in binary classification output layers.
#
# Advantages:
# - Output can be interpreted as probability.
#
# Disadvantage:
# - Can suffer from vanishing gradients.

# ReLU:
# ReLU(x) = max(0, x)
# Range: 0 to infinity
#
# Advantages:
# - Simple and computationally efficient.
# - Commonly used in hidden layers.
#
# Disadvantage:
# - Can suffer from "dying ReLU".

# Tanh:
# Range: -1 to 1
#
# Advantages:
# - Zero-centered.
# - Useful in some hidden-layer applications.
#
# Disadvantage:
# - Can suffer from vanishing gradients.

# Summary:
#
# Sigmoid → Binary output
# ReLU    → Common hidden-layer activation
# Tanh    → Zero-centered hidden activation


# ============================================================
# QUESTION 4
# Loss Function vs Cost Function
# ============================================================

# Loss Function:
# Measures the error for a single training example.

# Cost Function:
# Usually represents the average/aggregate loss over the training dataset.

# Example:
#
# Binary Cross-Entropy:
# Used for binary classification.
#
# Mean Squared Error (MSE):
# Commonly used for regression.

# Simple difference:
#
# Loss → Error for individual example
# Cost → Overall/average training error


# ============================================================
# QUESTION 5
# Optimizers
# ============================================================

# Optimizers update neural-network weights to minimize the loss function.

# 1. Gradient Descent:
# - Uses the gradient of the loss to update weights.
# - Simple but can be slow.
#
# 2. RMSprop:
# - Adapts the learning rate for each parameter.
# - Useful for problems with changing gradients.
#
# 3. Adam:
# - Combines ideas from Momentum and RMSprop.
# - Uses adaptive learning rates.
# - Usually converges quickly.
# - Common default choice for neural networks.

# Summary:
#
# Gradient Descent → Simple weight updates
# RMSprop         → Adaptive learning rates
# Adam            → Adaptive + momentum-based optimization

# QUESTION 6
# Single-Layer Perceptron from Scratch using NumPy
# Logical AND Gate
# ============================================================

import numpy as np

# AND gate data
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([0, 0, 0, 1])

# Initialize weights and bias
weights = np.zeros(2)
bias = 0

learning_rate = 0.1
epochs = 20


# Step activation function
def step_function(x):
    return 1 if x >= 0 else 0


# Training
for epoch in range(epochs):

    for i in range(len(X)):

        linear_output = np.dot(X[i], weights) + bias

        prediction = step_function(linear_output)

        error = y[i] - prediction

        weights += learning_rate * error * X[i]
        bias += learning_rate * error


# Test
print("Predictions:")

for x in X:
    result = step_function(np.dot(x, weights) + bias)
    print(x, "->", result)

# QUESTION 8
# Keras Neural Network on MNIST
# ============================================================

import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Flatten

# Load MNIST
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()

# Normalize
X_train = X_train / 255.0
X_test = X_test / 255.0


# Build model
model = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation="relu"),
    Dense(64, activation="relu"),
    Dense(10, activation="softmax")
])


# Compile
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)


# Train
history = model.fit(
    X_train,
    y_train,
    epochs=5,
    batch_size=32,
    validation_split=0.1
)


# Training accuracy
train_loss, train_accuracy = model.evaluate(
    X_train,
    y_train,
    verbose=0
)

print("Training Accuracy:", train_accuracy)

import tensorflow as tf

    tf.keras.datasets.fashion_mnist.load_data()

model = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation="relu"),
    Dense(64, activation="relu"),
    Dense(10, activation="softmax")
])


model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)


# Train
history = model.fit(
    X_train,
    y_train,
    epochs=10,
    batch_size=32,
    validation_split=0.1
)


# Loss curve
plt.figure(figsize=(8, 5))

plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Fashion MNIST Loss Curve")
plt.legend()
plt.grid()
plt.show()


# Accuracy curve
plt.figure(figsize=(8, 5))

plt.plot(history.history["accuracy"], label="Training Accuracy")
plt.plot(history.history["val_accuracy"], label="Validation Accuracy")

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Fashion MNIST Accuracy Curve")
plt.legend()
plt.grid()
plt.show()


# Interpretation:
#
# - Training loss should generally decrease.
# - Training accuracy should generally increase.
# - If training accuracy keeps increasing but validation accuracy decreases,
#   the model is likely overfitting.
# - If both training and validation accuracy remain low,
#   the model may be underfitting.
# - Similar training and validation curves indicate better generalization.

# STEP 8: Evaluation
# ------------------

# Important metrics:
#
# Precision:
# Of transactions predicted as fraud, how many were actually fraud?
#
# Recall:
# Of actual fraud transactions, how many were detected?
#
# F1-score:
# Balance between precision and recall.
#
# PR-AUC:
# Especially useful for highly imbalanced fraud datasets.
#
# ROC-AUC:
# Measures ranking/separation ability.

# Confusion Matrix:
#
#                 Predicted
#                 Legit  Fraud
# Actual Legit      TN      FP
# Actual Fraud      FN      TP


# For fraud detection:
# Recall is important because missing fraud (FN) causes financial loss.
#
# Precision is also important because too many false alarms (FP)
# can block legitimate customers.
