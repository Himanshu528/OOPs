# Theory

1. What is NumPy, and why is it widely used in Python?

NumPy is a Python library used for numerical and scientific computing.

It is widely used because it provides:
- Fast multidimensional arrays
- Mathematical operations
- Efficient memory usage
- Vectorized operations


2. How does broadcasting work in NumPy?

Broadcasting allows NumPy to perform operations on arrays with different shapes.

Example:
import numpy as np

a = np.array([1, 2, 3])
print(a + 10)

Output:
[11 12 13]


3. What is a Pandas DataFrame?

A DataFrame is a two-dimensional, labeled data structure in Pandas.

It contains:
- Rows
- Columns

Example:
import pandas as pd

df = pd.DataFrame({
    "Name": ["John", "Alice"],
    "Age": [25, 30]
})


4. Explain the use of the groupby() method in Pandas.

`groupby()` is used to group data based on one or more columns and perform operations on each group.

Example:
df.groupby("Department")["Salary"].mean()


5. Why is Seaborn preferred for statistical visualizations?

Seaborn is preferred because it:
- Provides attractive statistical plots
- Works directly with Pandas DataFrames
- Requires less code
- Provides built-in statistical functionality


6. Differences between NumPy arrays and Python lists.

NumPy Array:
- Faster for numerical operations
- Supports vectorized operations
- Uses less memory
- Supports multidimensional arrays

Python List:
- More flexible
- Can contain different data types
- Generally slower for numerical operations


7. What is a heatmap, and when should it be used?

A heatmap represents data using different colors.

It is commonly used to show:
- Correlations
- Patterns
- Relationships
- Matrix-like data


8. What does "vectorized operation" mean in NumPy?

A vectorized operation performs an operation on an entire array without explicitly using a Python loop.

Example:
import numpy as np

a = np.array([1, 2, 3])
print(a * 2)


9. How does Matplotlib differ from Plotly?

Matplotlib:
- Mainly creates static visualizations
- Highly customizable
- Commonly used for reports and basic plotting

Plotly:
- Creates interactive visualizations
- Supports zooming, hovering, and filtering
- Useful for dashboards and web applications


10. What is the significance of hierarchical indexing in Pandas?

Hierarchical indexing, or MultiIndex, allows multiple index levels in a DataFrame or Series.

It is useful for:
- Representing multi-level data
- Grouping data
- Complex data analysis


11. What is the role of Seaborn's pairplot() function?

`pairplot()` creates plots showing relationships between multiple numerical columns.

It is mainly used to:
- Find relationships
- Identify patterns
- Compare variables


12. What is the purpose of the describe() function in Pandas?

`describe()` provides summary statistics of numerical data.

It commonly returns:
- Count
- Mean
- Standard deviation
- Minimum
- Maximum
- Quartiles

Example:
df.describe()


13. Why is handling missing data important in Pandas?

Missing data can cause incorrect analysis and affect model results.

Pandas provides methods such as:
- `isnull()`
- `dropna()`
- `fillna()`


14. What are the benefits of using Plotly for data visualization?

- Interactive charts
- Zoom and hover functionality
- Easy dashboard integration
- Supports many chart types
- Works well with web applications


15. How does NumPy handle multidimensional arrays?

NumPy uses the `ndarray` object to handle multidimensional arrays.

Example:
import numpy as np

arr = np.array([[1, 2], [3, 4]])

print(arr)


16. What is the role of Bokeh in data visualization?

Bokeh is a Python library used to create interactive and web-based visualizations.

It is useful for:
- Interactive charts
- Dashboards
- Web applications


17. Difference between apply() and map() in Pandas.

`apply()`:
- Can be used on Series and DataFrames
- Applies a function across rows or columns

`map()`:
- Mainly used with Series
- Applies a function to each individual value


18. What are some advanced features of NumPy?

Some advanced features include:
- Broadcasting
- Vectorization
- Multidimensional arrays
- Linear algebra
- Random number generation
- Fast mathematical functions


19. How does Pandas simplify time series analysis?

Pandas provides tools for:
- Date and time handling
- Resampling
- Date-based indexing
- Rolling calculations
- Time-based grouping


20. What is the role of a pivot table in Pandas?

A pivot table summarizes and analyzes data by grouping it based on different columns.

Example:
pd.pivot_table(
    df,
    values="Salary",
    index="Department",
    aggfunc="mean"
)
