# QUESTION 1
# What is a random variable?

# A random variable is a variable whose value depends on the outcome
# of a random experiment.

# Example:
# Tossing a coin:
# X = 1 for Head
# X = 0 for Tail


# QUESTION 2
# Types of random variables

# 1. Discrete Random Variable
#    Takes countable values.
#    Example: Number of heads in 5 coin tosses.

# 2. Continuous Random Variable
#    Takes any value within a range.
#    Example: Height, weight, temperature.


# QUESTION 3
# Discrete vs Continuous Distribution

# Discrete:
# - Has countable values.
# - Example: Number of customers, number of defects.

# Continuous:
# - Can take any value within a range.
# - Example: Height, weight, time.


# QUESTION 4
# What is Binomial Distribution?

# Binomial distribution gives the probability of getting a certain
# number of successes in a fixed number of independent trials.

# Example:
# Tossing a coin 10 times and finding the probability of getting
# exactly 6 heads.

# Conditions:
# - Fixed number of trials.
# - Two possible outcomes.
# - Trials are independent.
# - Same probability of success.


# QUESTION 5
# What is Standard Normal Distribution?

# Standard normal distribution is a normal distribution with:
# Mean = 0
# Standard deviation = 1

# It is important because it allows us to calculate probabilities,
# z-scores, confidence intervals, and perform hypothesis testing.


# QUESTION 6
# What is Central Limit Theorem (CLT)?

# CLT states that when sufficiently large random samples are taken
# from a population, the distribution of sample means becomes
# approximately normal, even if the original population is not normal.

# Importance:
# - Helps estimate population means.
# - Used to calculate confidence intervals.
# - Important for statistical inference.


# QUESTION 7
# Significance of Confidence Intervals

# A confidence interval gives a range of values that is likely to
# contain the true population parameter.

# Example:
# A 95% confidence interval means that the method used to calculate
# the interval captures the true population parameter in about 95%
# of repeated samples.


# QUESTION 8
# Expected Value

# Expected value is the long-run average outcome of a random variable.

# Formula:
# E(X) = Σ [x * P(x)]

# Example:
# If a game gives:
# $10 with probability 0.5
# $20 with probability 0.5
#
# Expected value = (10 × 0.5) + (20 × 0.5) = $15


# QUESTION 9
# Generate 1000 random numbers from a normal distribution,
# calculate mean and standard deviation, and draw histogram.

import numpy as np
import matplotlib.pyplot as plt

numbers = np.random.normal(
    loc=50,
    scale=5,
    size=1000
)

mean = np.mean(numbers)
std = np.std(numbers)

print("Mean:", mean)
print("Standard Deviation:", std)

plt.hist(numbers, bins=30, edgecolor="black")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Normal Distribution")
plt.show()


# QUESTION 10
# Apply CLT and calculate 95% confidence interval for average sales.

import numpy as np

daily_sales = [
    220, 245, 210, 265, 230,
    250, 260, 275, 240, 255,
    235, 260, 245, 250, 225,
    270, 265, 255, 250, 260
]

# Calculate sample mean
mean_sales = np.mean(daily_sales)

# Calculate sample standard deviation
std_sales = np.std(daily_sales, ddof=1)

# Sample size
n = len(daily_sales)

# Standard error
standard_error = std_sales / np.sqrt(n)

# 95% confidence interval
# For a simple CLT-based calculation, use z = 1.96
z = 1.96

margin_of_error = z * standard_error

lower = mean_sales - margin_of_error
upper = mean_sales + margin_of_error

print("Mean Sales:", mean_sales)
print("Standard Deviation:", std_sales)
print("95% Confidence Interval:", (lower, upper))

# Interpretation:
# The estimated average daily sales is mean_sales.
# The 95% confidence interval gives a range of plausible values
# for the true average daily sales.
