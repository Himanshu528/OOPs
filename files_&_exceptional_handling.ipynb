# Theroy

1. Interpreted vs Compiled Languages
Interpreted: Code is executed by an interpreter, usually at runtime. Example: Python.
Compiled: Code is converted into machine code before execution. Example: C, C++.

2. What is exception handling in Python?

Exception handling is used to handle runtime errors without stopping the program.

Main keywords: try, except, else, finally.

3. Purpose of finally

finally always executes, whether an exception occurs or not.

It is commonly used for cleanup, such as closing files.

4. What is logging in Python?

Logging is the process of recording application events, errors, and information for debugging and monitoring.

5. Significance of __del__

__del__ is called when an object is being destroyed.

It can be used for cleanup, but should not be relied on for critical resource management.

6. import vs from ... import
import module → Imports the whole module.
from module import function → Imports a specific function/class.
import math
math.sqrt(25)

from math import sqrt
sqrt(25)

7. How to handle multiple exceptions?

Use multiple except blocks or handle multiple exception types in one block.

try:
    x = 10 / 0
except (ZeroDivisionError, ValueError):
    print("Error")

8. Purpose of with when handling files

with automatically closes the file after use, even if an exception occurs.

with open("file.txt") as file:
    data = file.read()

9. Multithreading vs Multiprocessing
Multithreading: Multiple threads within the same process; useful for I/O tasks.
Multiprocessing: Multiple processes; useful for CPU-intensive tasks.

10. Advantages of logging
Helps debug errors
Monitors application behavior
Records important events
Helps troubleshoot production issues

11. What is memory management in Python?

Memory management is how Python allocates, uses, and releases memory for objects.

Python handles most memory management automatically.

12. Basic steps in exception handling
Put risky code inside try.
Handle errors using except.
Use else for code when no exception occurs.
Use finally for cleanup.
13. Why is memory management important?

It helps:

Avoid unnecessary memory usage
Prevent memory leaks
Improve application performance

14. Role of try and except
try → Contains code that may cause an exception.
except → Handles the exception.
try:
    x = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero")

15. How does Python garbage collection work?

Python automatically removes objects that are no longer reachable.

It mainly uses:

Reference counting
Garbage collector for cyclic references

16. Purpose of else in exception handling

else executes only when no exception occurs in the try block.

try:
    x = 10 / 2
except:
    print("Error")
else:
    print("Success")

17. Common logging levels

From low to high severity:

DEBUG
INFO
WARNING
ERROR
CRITICAL

18. os.fork() vs multiprocessing
os.fork() → Creates a child process using the OS-level fork mechanism; mainly available on Unix-like systems.
multiprocessing → Python's higher-level API for creating and managing processes; cross-platform.

19. Importance of closing a file

Closing a file:

Releases system resources
Ensures buffered data is written
Prevents file/resource leaks

Using with handles this automatically.

20. file.read() vs file.readline()
read() → Reads the entire file or specified number of characters.
readline() → Reads one line at a time.

21. What is the logging module used for?

Python's logging module is used to create and manage log messages from an application.

import logging

logging.info("Application started")

22. What is the os module used for in file handling?

The os module provides functions for interacting with the operating system and files/directories.

Examples:

os.mkdir()
os.remove()
os.rename()
os.listdir()

23. Challenges associated with memory management

Common challenges include:

Memory leaks
High memory usage
Circular references
Large objects consuming memory
Unnecessary object creation

24. How do you raise an exception manually in Python?

Use the `raise` keyword to manually raise an exception.

Example:
raise ValueError("Invalid value")


25. Why is it important to use multithreading in certain applications?

Multithreading allows multiple tasks to run concurrently.

It is useful for:
- I/O operations
- Network requests
- File handling
- Improving responsiveness of applications

43. How would you implement a basic logging setup that logs to a file with rotation after 1MB?

Use `RotatingFileHandler`.

Example:
import logging
from logging.handlers import RotatingFileHandler

handler = RotatingFileHandler(
    "app.log",
    maxBytes=1024 * 1024,
    backupCount=3
)

logging.basicConfig(
    handlers=[handler],
    level=logging.INFO
)

logging.info("Application started")
