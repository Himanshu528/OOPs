# QUESTION 1
# What is hypothesis testing?

# Hypothesis testing is a statistical method used to determine
# whether there is enough evidence to support a claim about a population.


# QUESTION 2
# Null Hypothesis vs Alternative Hypothesis

# Null Hypothesis (H0):
# Assumes there is no significant effect or difference.

# Alternative Hypothesis (H1):
# Assumes there is a significant effect or difference.

# Example:
# H0: Average salary = 50,000
# H1: Average salary != 50,000


# QUESTION 3
# Significance Level

# Significance level (alpha) is the probability of rejecting the
# null hypothesis when it is actually true.

# Common value:
# alpha = 0.05

# Decision:
# p-value < 0.05  -> Reject H0
# p-value >= 0.05 -> Fail to reject H0


# QUESTION 4
# Type I and Type II Errors

# Type I Error:
# Rejecting a true null hypothesis.
# Example: A medical test says a healthy person has a disease.

# Type II Error:
# Failing to reject a false null hypothesis.
# Example: A medical test says a sick person is healthy.


# QUESTION 5
# Z-Test vs T-Test

# Z-Test:
# - Used when population standard deviation is known.
# - Generally used for large samples.
# - Uses the standard normal distribution.

# T-Test:
# - Used when population standard deviation is unknown.
# - Commonly used for small samples.
# - Uses the t-distribution.

# QUESTION 7
# Hypothesis Testing using Z-Statistics

import numpy as np
from scipy.stats import norm

sample_data = [
    49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7,
    50.2, 49.6, 50.1, 49.9, 50.8, 50.4, 48.9, 50.6,
    50.0, 49.7, 50.2, 49.5, 50.1, 50.3, 50.4, 50.5,
    50.0, 50.7, 49.3, 49.8, 50.2, 50.9, 50.3, 50.4,
    50.0, 49.7, 50.5, 49.9
]

# Assumed population mean
population_mean = 50

# Known population standard deviation
population_std = 1

sample_mean = np.mean(sample_data)
n = len(sample_data)

# Z-statistic
z_score = (sample_mean - population_mean) / (
    population_std / np.sqrt(n)
)

# Two-tailed p-value
p_value = 2 * (1 - norm.cdf(abs(z_score)))

print("Sample Mean:", sample_mean)
print("Z-Score:", z_score)
print("P-Value:", p_value)

if p_value < 0.05:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")

# Interpretation:
# If p-value < 0.05, there is significant evidence that
# the population mean is different from 50.
#
# Otherwise, there is not enough evidence to say that
# the population mean is different from 50.


# QUESTION 8
# Simulate Normal Data and Calculate 95% Confidence Interval

import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(
    loc=50,
    scale=5,
    size=100
)

mean = np.mean(data)
std = np.std(data, ddof=1)
n = len(data)

standard_error = std / np.sqrt(n)

# 95% Confidence Interval
margin_of_error = 1.96 * standard_error

lower = mean - margin_of_error
upper = mean + margin_of_error

print("Mean:", mean)
print("95% Confidence Interval:", (lower, upper))

plt.hist(data, bins=20, edgecolor="black")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Normal Distribution")
plt.show()


# QUESTION 9
# Calculate and visualize Z-Scores

import numpy as np
import matplotlib.pyplot as plt

def calculate_z_scores(data):
    mean = np.mean(data)
    std = np.std(data)

    z_scores = (data - mean) / std

    return z_scores


data = [10, 12, 15, 18, 20, 22, 25, 30]

z_scores = calculate_z_scores(data)

print("Original Data:", data)
print("Z-Scores:", z_scores)

plt.hist(z_scores, bins=8, edgecolor="black")
plt.xlabel("Z-Score")
plt.ylabel("Frequency")
plt.title("Standardized Data")
plt.show()

# Interpretation:
# Z-score tells how many standard deviations a value is
# away from the mean.
#
# Z = 0  -> Value equals the mean.
# Z = +1 -> 1 standard deviation above the mean.
# Z = -1 -> 1 standard deviation below the mean.
