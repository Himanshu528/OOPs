# Theroy

1. What is Python, and why is it popular?

Python is a high-level, interpreted programming language.

It is popular because it:
- Has simple and readable syntax
- Is easy to learn
- Has many libraries and frameworks
- Supports multiple programming paradigms
- Is widely used in Web Development, AI, Data Science, and Automation


2. What is an interpreter in Python?

An interpreter executes Python code and converts it into instructions that the computer can execute.

It allows Python code to be executed without manually compiling it first.


3. What are pre-defined keywords in Python?

Keywords are reserved words that have a special meaning in Python.

Examples:
if, else, for, while, class, def, return, True, False


4. Can keywords be used as variable names?

No. Python keywords cannot be used as variable names.

Example:
class = 10   # Invalid


5. What is mutability in Python?

Mutability means whether an object's value can be changed after it is created.

- Mutable → Can be changed
- Immutable → Cannot be changed

Example:
List → Mutable
Tuple → Immutable


6. Why are lists mutable, but tuples are immutable?

Lists are designed to allow modification of their elements, while tuples are designed to remain unchanged after creation.

Example:
list1 = [1, 2, 3]
list1[0] = 10

tuple1 = (1, 2, 3)
# tuple1[0] = 10  → Error


7. Difference between == and is operators.

`==` checks whether two objects have the same value.

`is` checks whether two variables refer to the same object in memory.

Example:
a = [1, 2]
b = [1, 2]

a == b   # True
a is b   # False


8. What are logical operators in Python?

Logical operators are used to combine conditions.

- and → Both conditions must be True
- or → At least one condition must be True
- not → Reverses the result


9. What is type casting in Python?

Type casting is the process of converting one data type into another.

Example:
x = "10"
y = int(x)


10. Difference between implicit and explicit type casting.

Implicit casting → Python automatically converts the data type.

Example:
x = 10
y = 2.5
result = x + y

Explicit casting → Programmer manually converts the data type.

Example:
x = "10"
y = int(x)


11. Purpose of conditional statements in Python.

Conditional statements are used to execute different blocks of code based on conditions.

Examples:
- if
- elif
- else


12. How does the elif statement work?

`elif` checks another condition when the previous `if` or `elif` condition is False.

Example:
if age >= 18:
    print("Adult")
elif age >= 13:
    print("Teenager")
else:
    print("Child")


13. Difference between for and while loops.

for loop:
- Used when the number of iterations is known or when iterating over a sequence.

while loop:
- Used when the number of iterations is not known and depends on a condition.


14. Scenario where a while loop is more suitable than a for loop.

A while loop is suitable when we don't know how many times the loop should run.

Example:

password = ""

while password != "1234":
    password = input("Enter password: ")

print("Access granted")
