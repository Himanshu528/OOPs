# Theory

1. What are the key differences between SQL and NoSQL databases?

SQL:
- Relational databases with tables and rows.
- Fixed/predefined schema.
- Uses SQL queries.
- Supports strong relationships and JOINs.
- Examples: MySQL, PostgreSQL.

NoSQL:
- Non-relational databases.
- Flexible schema.
- Stores data as documents, key-value, graphs, etc.
- Designed for scalability and large distributed data.
- Examples: MongoDB, Redis.


2. What makes MongoDB a good choice for modern applications?

- Flexible document-based schema.
- Easy to handle changing data structures.
- High scalability using sharding.
- High availability using replication.
- Supports powerful aggregation.
- Good for large and rapidly changing data.


3. Explain the concept of collections in MongoDB.

A collection is a group of MongoDB documents.

It is similar to a table in SQL.

Example:
db.users.insertOne({
    name: "John",
    age: 25
})

Here, "users" is a collection.


4. How does MongoDB ensure high availability using replication?

MongoDB uses Replica Sets.

- One Primary node handles writes.
- Secondary nodes maintain copies of the data.
- If the Primary fails, a Secondary can become the new Primary.
- This provides high availability and data redundancy.


5. What are the main benefits of MongoDB Atlas?

- Fully managed MongoDB service.
- Automatic backups.
- Easy scaling.
- Built-in security.
- Monitoring and performance tools.
- High availability.
- Available on AWS, Azure, and Google Cloud.


6. What is the role of indexes in MongoDB?

Indexes improve query performance by allowing MongoDB to find documents faster without scanning the entire collection.

Example:

db.users.createIndex({ email: 1 })

This creates an index on the email field.


7. Describe the stages of the MongoDB aggregation pipeline.

Common stages include:

- $match → Filters documents.
- $group → Groups documents and performs calculations.
- $sort → Sorts documents.
- $project → Selects or modifies fields.
- $lookup → Joins data from another collection.
- $unwind → Deconstructs an array.
- $limit → Limits the number of documents.


8. What is sharding in MongoDB? How does it differ from replication?

Sharding:
- Distributes data across multiple servers.
- Used for horizontal scaling.
- Helps handle very large datasets and high traffic.

Replication:
- Creates copies of the same data on multiple servers.
- Used mainly for high availability and redundancy.

Sharding = distributes data.
Replication = copies data.


9. What is PyMongo, and why is it used?

PyMongo is the official Python driver for MongoDB.

It allows Python applications to:
- Connect to MongoDB.
- Insert documents.
- Read documents.
- Update documents.
- Delete documents.
- Run MongoDB queries.


10. What are the ACID properties in MongoDB transactions?

ACID means:

- Atomicity → Transaction completes fully or not at all.
- Consistency → Data remains valid.
- Isolation → Transactions do not interfere incorrectly.
- Durability → Committed data is preserved.

MongoDB supports ACID transactions for multiple documents.


11. What is the purpose of MongoDB's explain() function?

explain() shows how MongoDB executes a query.

It helps analyze:
- Query performance.
- Index usage.
- Documents examined.
- Execution time.

Example:

db.users.find({ email: "john@gmail.com" }).explain("executionStats")


12. How does MongoDB handle schema validation?

MongoDB allows optional schema validation using JSON Schema.

It can enforce rules such as:
- Required fields.
- Data types.
- Minimum/maximum values.

Example:

{
    $jsonSchema: {
        required: ["name", "email"]
    }
}


13. What is the difference between a primary and a secondary node in a replica set?

Primary:
- Accepts write operations.
- Usually handles read operations as well.
- Replicates data to secondary nodes.

Secondary:
- Copies data from the Primary.
- Can be used for certain read operations.
- Can become Primary if the current Primary fails.


14. What security mechanisms does MongoDB provide for data protection?

MongoDB provides:

- Authentication.
- Authorization using roles.
- Encryption in transit using TLS.
- Encryption at rest.
- Network access controls.
- Auditing.
- Role-Based Access Control (RBAC).


15. Explain embedded documents and when they should be used.

Embedded documents store related data inside the same document.

Example:

{
    name: "John",
    address: {
        city: "Delhi",
        country: "India"
    }
}

Use embedded documents when:
- Data is closely related.
- Data is usually accessed together.
- The embedded data does not need independent querying.


16. What is the purpose of the $lookup stage?

$lookup is used to combine data from two MongoDB collections.

It is similar to a JOIN in SQL.

Example:

db.orders.aggregate([
    {
        $lookup: {
            from: "customers",
            localField: "customerId",
            foreignField: "_id",
            as: "customer"
        }
    }
])


17. What are some common use cases for MongoDB?

- Web and mobile applications.
- Real-time applications.
- Content management systems.
- Product catalogs.
- IoT applications.
- Analytics.
- Social media applications.
- Applications with frequently changing data.


18. What are the advantages of MongoDB for horizontal scaling?

- Uses sharding to distribute data.
- Can add more servers as data grows.
- Handles high traffic.
- Distributes workload across multiple servers.
- Supports large datasets.


19. How do MongoDB transactions differ from SQL transactions?

SQL:
- Transactions are commonly used for relational operations across multiple tables.
- Strong relationships and JOINs are common.

MongoDB:
- Single-document operations are atomic by default.
- Multi-document transactions are also supported.
- Transactions can work across multiple documents and collections.

MongoDB generally encourages embedding related data when appropriate to reduce the need for transactions.


20. What are the main differences between capped collections and regular collections?

Capped Collection:
- Fixed maximum size.
- Maintains insertion order.
- Automatically removes old documents when the size limit is reached.
- Useful for logs and real-time data.

Regular Collection:
- No fixed size limit by default.
- Documents can be inserted, updated, and deleted normally.
- Suitable for general application data.


21. What is the purpose of the $match stage?

$match filters documents based on specified conditions.

It is similar to the WHERE clause in SQL.

Example:

db.products.aggregate([
    {
        $match: {
            price: { $gt: 1000 }
        }
    }
])


22. How can you secure access to a MongoDB database?

- Enable authentication.
- Create users with appropriate roles.
- Use strong passwords.
- Enable TLS/SSL encryption.
- Restrict network access.
- Use firewalls/security groups.
- Enable encryption at rest.
- Follow the principle of least privilege.


23. What is MongoDB's WiredTiger storage engine, and why is it important?

WiredTiger is MongoDB's default storage engine.

It provides:
- Document-level concurrency.
- Compression.
- Efficient memory usage.
- Crash recovery.
- Journaling.
- Good read/write performance.

It is important because it provides reliable and efficient data storage for MongoDB applications.

            "_id": "$Region",

            "TotalSales": -1
