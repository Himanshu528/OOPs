# Statistics Basics

# QUESTION 1
# Descriptive vs Inferential Statistics

# Descriptive statistics:
# Summarizes and describes the collected data.
# Example: Average salary of employees in a company.

# Inferential statistics:
# Uses sample data to make conclusions or predictions about a larger population.
# Example: Surveying 1,000 customers to estimate the satisfaction of all customers.


# QUESTION 2
# Sampling

# Sampling is the process of selecting a subset of a population for analysis.

# Random Sampling:
# Every member of the population has an equal chance of being selected.

# Stratified Sampling:
# Population is divided into groups (strata), and samples are taken from each group.

# Example:
# Random -> Select 100 employees randomly.
# Stratified -> Select employees separately from HR, IT, Sales, etc.


# QUESTION 3
# Mean, Median, and Mode

# Mean:
# Sum of all values divided by the number of values.
# Example: [10, 20, 30] -> Mean = 20

# Median:
# Middle value after sorting the data.
# Example: [10, 20, 30] -> Median = 20

# Mode:
# Most frequently occurring value.
# Example: [10, 20, 20, 30] -> Mode = 20

# They are important because they help understand the central
# or typical value of a dataset.


# QUESTION 4
# Skewness and Kurtosis

# Skewness:
# Measures the asymmetry of data distribution.

# Positive skew:
# Data has a longer tail on the right side.
# Usually: Mean > Median.

# Kurtosis:
# Measures the heaviness of the tails and the peakedness of a distribution.


# QUESTION 5

import statistics

# QUESTION 6
# Calculate Covariance and Correlation

import numpy as np

list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]

covariance = np.cov(list_x, list_y)[0, 1]
correlation = np.corrcoef(list_x, list_y)[0, 1]

print("Covariance:", covariance)
print("Correlation:", correlation)

# Positive covariance -> both variables tend to increase together.
# Correlation close to +1 -> strong positive linear relationship.

# QUESTION 8
# Advertising Spend vs Daily Sales

# Covariance:
# Shows the direction of the relationship.
# Positive covariance means advertising spend and sales tend
# to increase together.

# Correlation:
# Measures the strength and direction of the linear relationship.
# Value ranges from -1 to +1.

advertising_spend = [200, 250, 300, 400, 500]
daily_sales = [2200, 2450, 2750, 3200, 4000]

correlation = np.corrcoef(
    advertising_spend,
    daily_sales
)[0, 1]

print("Correlation:", correlation)

# A value close to +1 indicates a strong positive relationship.


# QUESTION 9
# Customer Satisfaction Analysis

# Useful summary statistics:
# Mean -> Average satisfaction score.
# Median -> Middle satisfaction score.
# Mode -> Most common score.
# Standard deviation -> Shows how spread out the scores are.
# Minimum/Maximum -> Shows the range of scores.

# Useful visualization:
# Histogram -> Shows the distribution of survey scores.
# Boxplot -> Helps identify spread and outliers.

survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]

print("Mean:", np.mean(survey_scores))
print("Median:", np.median(survey_scores))
print("Standard Deviation:", np.std(survey_scores))

plt.hist(survey_scores, bins=range(1, 12), edgecolor="black")
plt.xlabel("Survey Score")
plt.ylabel("Frequency")
plt.title("Customer Satisfaction Scores")
plt.show()
