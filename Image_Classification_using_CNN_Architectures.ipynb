#  Image Classification using CNN Architectures

# ============================================================
# QUESTION 1
# What is a CNN?
# ============================================================

# CNN (Convolutional Neural Network) is a deep learning model mainly used
# for image and visual data.

# CNN architecture:
#
# Image
#   ↓
# Convolution
#   ↓
# ReLU
#   ↓
# Pooling
#   ↓
# Convolution
#   ↓
# Fully Connected Layer
#   ↓
# Output

# CNN vs Fully Connected Network:
#
# CNN:
# - Uses convolution filters to detect local patterns.
# - Preserves spatial relationships between pixels.
# - Uses fewer parameters because filters are shared.
# - Highly effective for image classification.
#
# Fully Connected Network:
# - Every neuron connects to every input.
# - Usually requires flattening the image.
# - Has many more parameters for images.
# - Can lose spatial relationships.
#
# CNN is generally more efficient and accurate for image data.


# ============================================================
# QUESTION 2
# LeNet-5 Architecture
# ============================================================

# LeNet-5 was introduced by Yann LeCun and colleagues for handwritten
# digit recognition.

# Architecture:
#
# Input: 32 × 32
#      ↓
# C1: Convolution
#      ↓
# S2: Average Pooling
#      ↓
# C3: Convolution
#      ↓
# S4: Average Pooling
#      ↓
# C5: Convolution
#      ↓
# F6: Fully Connected
#      ↓
# Output: 10 classes
#
# Key ideas:
# - Convolution layers learn image features.
# - Pooling reduces spatial dimensions.
# - Fully connected layers perform classification.
#
# Importance:
# - Demonstrated that CNNs can automatically learn visual features.
# - Became an important foundation for modern computer vision.
# - Inspired later CNN architectures such as AlexNet and VGG.

# Original research paper:
#
# Y. LeCun, L. Bottou, Y. Bengio, P. Haffner,
# "Gradient-Based Learning Applied to Document Recognition",
# Proceedings of the IEEE, 1998.
#
# The paper introduced the LeNet-style architecture and its application
# to document/handwritten digit recognition.


# ============================================================
# QUESTION 3
# AlexNet vs VGGNet
# ============================================================

# AlexNet:
# - Introduced in 2012.
# - Around 60 million parameters.
# - Used 5 convolutional layers + 3 fully connected layers.
# - Used ReLU activation.
# - Used Dropout to reduce overfitting.
# - Used GPU training.
# - Used data augmentation.
#
# Key innovation:
# ReLU + GPU training + Dropout + deep CNN architecture.
#
# Limitation:
# - Large number of parameters.
# - Fully connected layers are computationally expensive.


# VGGNet:
# - Introduced in 2014.
# - Common versions: VGG16 and VGG19.
# - VGG16 has about 138 million parameters.
# - Uses many small 3×3 convolution filters.
# - Has a very simple and uniform architecture.
#
# Key innovation:
# Repeated 3×3 convolutions create deeper networks while maintaining
# a simple design.
#
# Limitations:
# - Very large number of parameters.
# - High memory usage.
# - Computationally expensive.

# Comparison:
#
# AlexNet → Earlier, less deep, ~60M parameters
# VGG16   → Deeper, ~138M parameters, simpler repeated 3×3 filters


# ============================================================
# QUESTION 4
# Transfer Learning
# ============================================================

# Transfer learning uses a model that has already been trained on a large
# dataset and adapts it to a new task.

# Example:
#
# ImageNet-trained ResNet/VGG
#          ↓
# Remove original classifier
#          ↓
# Add new classifier
#          ↓
# Train on custom dataset

# Advantages:
# - Requires less training data.
# - Reduces training time.
# - Reduces computational cost.
# - Usually gives better performance on small datasets.
# - Pre-trained layers already contain useful features such as edges,
#   textures and shapes.

# Typical approach:
# 1. Freeze most pre-trained layers.
# 2. Replace the final classification layer.
# 3. Train the new classifier.
# 4. Optionally unfreeze some deeper layers.
# 5. Fine-tune with a small learning rate.


# ============================================================
# QUESTION 5
# Residual Connections in ResNet
# ============================================================

# A residual connection creates a shortcut around one or more layers.

# Normal network:
#
# Input → Layer → Layer → Output
#
# ResNet:
#
# Input ─────────────────────┐
#   ↓                        │
# Layer → Layer → F(x) ── + ─┘
#                          ↓
#                      ReLU
#
# Instead of directly learning H(x), the network learns:
#
# F(x) = H(x) - x
#
# Output:
# F(x) + x

# Benefits:
# - Provides a shortcut path for gradients.
# - Reduces vanishing-gradient problems.
# - Makes very deep networks easier to train.
# - Helps preserve useful information.
#
# This allowed ResNet to successfully train networks with very large
# numbers of layers.

# QUESTION 6
# LeNet-5 using TensorFlow + MNIST
# ============================================================

import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import (
    Conv2D,
    AveragePooling2D,
    Flatten,
    Dense
)
import time

# Load MNIST
(X_train, y_train), (X_test, y_test) = \
    tf.keras.datasets.mnist.load_data()

# Normalize
X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

# Add channel dimension
X_train = X_train[..., None]
X_test = X_test[..., None]

# LeNet-style model
model = Sequential([
    Conv2D(6, kernel_size=5, activation="tanh",
           padding="same", input_shape=(28, 28, 1)),

    AveragePooling2D(pool_size=2),

    Conv2D(16, kernel_size=5, activation="tanh"),

    AveragePooling2D(pool_size=2),

    Flatten(),

    Dense(120, activation="tanh"),

    Dense(84, activation="tanh"),

    Dense(10, activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

start_time = time.time()

history = model.fit(
    X_train,
    y_train,
    epochs=5,
    batch_size=128,
    validation_split=0.1,
    verbose=1
)

training_time = time.time() - start_time

test_loss, test_accuracy = model.evaluate(
    X_test,
    y_test,
    verbose=0
)

print("Test Accuracy:", test_accuracy)
print("Training Time:", training_time, "seconds")

# Note:
# Exact accuracy and training time depend on the Colab hardware,
# TensorFlow version and training configuration.
# MNIST accuracy should generally be very high (around 98%+).

# QUESTION 7
# VGG16 Transfer Learning
# Small Custom Dataset
# ============================================================

# Dataset structure:
#
# flowers/
#   rose/
#   sunflower/
#   tulip/
#   daisy/
#
# Place the dataset in /content/flowers in Google Colab.

import tensorflow as tf
from tensorflow.keras.applications import VGG16
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D
from tensorflow.keras.models import Model
from tensorflow.keras.preprocessing.image import ImageDataGenerator

IMG_SIZE = (224, 224)
BATCH_SIZE = 32

datagen = ImageDataGenerator(
    rescale=1./255,
    validation_split=0.2
)

train_data = datagen.flow_from_directory(
    "/content/flowers",
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    subset="training"
)

val_data = datagen.flow_from_directory(
    "/content/flowers",
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    subset="validation"
)

# Load pre-trained VGG16
base_model = VGG16(
    weights="imagenet",
    include_top=False,
    input_shape=(224, 224, 3)
)

# Freeze base layers initially
base_model.trainable = False

x = base_model.output
x = GlobalAveragePooling2D()(x)

x = Dense(128, activation="relu")(x)

output = Dense(
    train_data.num_classes,
    activation="softmax"
)(x)

model = Model(
    inputs=base_model.input,
    outputs=output
)

model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

# Train classifier
history = model.fit(
    train_data,
    validation_data=val_data,
    epochs=5
)

# Fine-tuning
base_model.trainable = True

# Freeze earlier layers and fine-tune deeper layers
for layer in base_model.layers[:-4]:
    layer.trainable = False

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=1e-5
    ),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

history_fine = model.fit(
    train_data,
    validation_data=val_data,
    epochs=5
)

# Result discussion:
#
# - Transfer learning reduces training time.
# - Pre-trained VGG16 already knows useful visual features.
# - Initial freezing prevents destroying learned ImageNet features.
# - Fine-tuning deeper layers adapts the model to flower/animal features.
# - Compare training and validation accuracy to check generalization.
# - If training accuracy is high but validation accuracy is much lower,
#   overfitting is occurring.

import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np

from tensorflow.keras.applications import AlexNet

# Note:
# AlexNet is not available in all TensorFlow/Keras versions.
# If unavailable, use a compatible AlexNet implementation or
# tf.keras.applications with a similar CNN architecture.

# Example compatible approach using a small AlexNet-style model:

alexnet = tf.keras.Sequential([
    tf.keras.layers.Conv2D(
        96, (11, 11),
        strides=4,
        activation="relu",
        input_shape=(224, 224, 3),
        name="conv1"
    ),
    tf.keras.layers.MaxPooling2D((3, 3), strides=2),
    tf.keras.layers.Conv2D(
        256, (5, 5),
        padding="same",
        activation="relu"
    ),
    tf.keras.layers.MaxPooling2D((3, 3), strides=2),
])

# Example input image
image = np.random.random((1, 224, 224, 3)).astype("float32")

# Extract first convolutional layer
conv_layer = alexnet.get_layer("conv1")

# Create feature-map model
feature_model = tf.keras.Model(
    inputs=alexnet.input,
    outputs=conv_layer.output
)

feature_maps = feature_model.predict(image)

# Visualize first 16 feature maps
plt.figure(figsize=(12, 8))

for i in range(16):
    plt.subplot(4, 4, i + 1)
    plt.imshow(feature_maps[0, :, :, i], cmap="gray")
    plt.axis("off")
    plt.title(f"Filter {i + 1}")

plt.tight_layout()
plt.show()

# Interpretation:
# Early convolution filters generally learn simple patterns such as:
# - Edges
# - Lines
# - Textures
# - Color transitions
#
# Feature maps show where those learned patterns are activated
# in the input image.

# QUESTION 9
# GoogLeNet / Inception-style CNN on CIFAR-10
# ============================================================

import tensorflow as tf
import matplotlib.pyplot as plt

from tensorflow.keras import layers, Model

# Load CIFAR-10
(X_train, y_train), (X_test, y_test) = \
    tf.keras.datasets.cifar10.load_data()

X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

# Simple Inception block
def inception_block(x, filters):

    branch1 = layers.Conv2D(
        filters, 1, padding="same", activation="relu"
    )(x)

    branch3 = layers.Conv2D(
        filters, 1, padding="same", activation="relu"
    )(x)

    branch3 = layers.Conv2D(
        filters, 3, padding="same", activation="relu"
    )(branch3)

    branch5 = layers.Conv2D(
        filters, 1, padding="same", activation="relu"
    )(x)

    branch5 = layers.Conv2D(
        filters, 5, padding="same", activation="relu"
    )(branch5)

    pool = layers.MaxPooling2D(
        3, strides=1, padding="same"
    )(x)

    pool = layers.Conv2D(
        filters, 1, padding="same", activation="relu"
    )(pool)

    return layers.Concatenate()([
        branch1,
        branch3,
        branch5,
        pool
    ])


inputs = layers.Input(shape=(32, 32, 3))

x = layers.Conv2D(
    32, 3, padding="same", activation="relu"
)(inputs)

x = inception_block(x, 32)

x = layers.MaxPooling2D()(x)

x = inception_block(x, 64)

x = layers.GlobalAveragePooling2D()(x)

outputs = layers.Dense(
    10,
    activation="softmax"
)(x)

model = Model(inputs, outputs)

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

history = model.fit(
    X_train,
    y_train,
    epochs=10,
    batch_size=64,
    validation_split=0.1
)

# Plot training and validation accuracy

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("CIFAR-10 Training vs Validation Accuracy")
plt.legend()
plt.grid()
plt.show()

# Analysis:
#
# If training accuracy increases and validation accuracy also increases:
# → Model is learning well.
#
# If training accuracy keeps increasing but validation accuracy stops
# increasing or decreases:
# → Overfitting.
#
# If both training and validation accuracy remain low:
# → Underfitting.
#
# Possible solutions for overfitting:
# - Dropout
# - Data augmentation
# - L2 regularization
# - Early stopping
